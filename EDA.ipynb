### Diagnóstico y Plan de Tratamiento de Valores Ausentes

Tras inspeccionar las 51,525 filas del dataset, se identificaron valores nulos en 5 columnas:

1. **is_4wd (50.37%):** Los nulos representan la ausencia de tracción 4x4; se imputan con `0.0`.
2. **paint_color (17.99%):** Al ser categórica, se imputan con `'unknown'` para no perder información de otras variables.
3. **odometer (15.32%) y cylinders (10.21%):** Se imputan utilizando la mediana de cada variable agrupada por el modelo del vehículo.
4. **model_year (7.02%):** Se imputa con la mediana por modelo para preservar la totalidad de los datos.

### Hallazgos de Calidad de Datos y Casos Atípicos

1. **Inconsistencias en Precios:** Se detectaron 863 anuncios (1.67%) con precios inferiores a $100 (incluyendo valores de $1). Se establece la regla de filtrar `price >= 100` en las visualizaciones de la aplicación interactiva para mantener la precisión de la distribución.
2. **Velocidad de Venta (days_listed):** Un total de 54 vehículos (0.10%) se registran con 0 días de publicación, lo cual refleja ventas inmediatas o cancelaciones el mismo día de carga. Se conservan por ser registros legítimos del comportamiento del mercado.

## TRATAMIENTO Y CORRECCIÓN DE DATOS

```python
# 1. Imputación de variables categóricas y booleanas
df['is_4wd'] = df['is_4wd'].fillna(0.0)
df['paint_color'] = df['paint_color'].fillna('unknown')

# 2. Imputación de numéricas utilizando la mediana agrupada por modelo de vehículo
df['model_year'] = df['model_year'].fillna(df.groupby('model')['model_year'].transform('median'))
df['cylinders'] = df['cylinders'].fillna(df.groupby('model')['cylinders'].transform('median'))
df['odometer'] = df['odometer'].fillna(df.groupby('model')['odometer'].transform('median'))

# 3. Filtrado de publicaciones con precios atípicos (inconsistentes)
df_cleaned = df[df['price'] >= 100].copy()

# 4. Verificación final de valores nulos en el DataFrame limpio
print("--- Conteo de nulos tras la limpieza ---")
print(df_cleaned.isna().sum())
print(f"\nTotal de filas resultantes: {len(df_cleaned)}")
```

```python
# Imputación de la variable 'odometer' con la mediana global
df_cleaned['odometer'] = df_cleaned['odometer'].fillna(df_cleaned['odometer'].median())

df_cleaned.info()
```

## ANÁLISIS EXPLORATORIO DE DATOS (EDA)

## Histograma: Distribución de Precios

```python
# Histograma para analizar la distribución del precio
fig_hist = px.histogram(
    df_cleaned, 
    x='price',
    nbins=50,
    title='Distribución del Precio de los Vehículos',
    labels={'price': 'Precio del vehículo ($)'}
)

# Cambiar la etiqueta del eje Y
fig_hist.update_layout(yaxis_title='Cantidad de anuncios')

fig_hist.show()
```

### Comentarios respecto a la Distribución del Precio (Histograma)


* Eje X (Precio del vehículo): Muestra los rangos de costo a los que los vendedores ofertan sus autos.
* Eje Y (Cantidad de anuncios): Indica cuántos anuncios o publicaciones existen en la plataforma para cada rango de precio.

**Hallazgos principales:**
  1. Concentración en precios bajos: La barra más alta se ubica al principio del gráfico. Esto significa que la gran mayoría de las publicaciones ofertan vehículos económicos (concentrados principalmente entre $100 y $15,000 dólares).
  2. Caída rápida de la oferta: Conforme el precio aumenta, la cantidad de anuncios disminuye de forma drástica. 
  3. Valores poco comunes: Hay muy pocos vehículos con precios superiores a $50,000 dólares y casos contados que superan los $100,000 (autos de lujo, de trabajo o piezas de colección)[cite: 1].

* Podríamos concluir que el mercado de autos usados en esta plataforma, está dominado por vehículos accesibles de gama media y baja, con muy pocas opciones de precios elevados.

# Gráfico de Dispersión: Precio vs. Kilometraje

```python
# Gráfico de dispersión para analizar la depreciación por kilometraje
fig_scatter = px.scatter(
    df_cleaned, 
    x='odometer', 
    y='price', 
    title='Relación entre Kilometraje y Precio'
)
fig_scatter.show()
```

### Comentarios rspecto a la Relación entre Kilometraje y Precio (Gráfico de Dispersión)

  * Cada punto azul representa **un vehículo individual** publicado en la plataforma[cite: 1, 2].
  * Eje X (Odometer / Kilometraje): Indica el uso que tiene el auto medido en millas[cite: 1, 3].
  * Eje Y (Price / Precio): Indica el valor en dólares al que se está vendiendo[cite: 1, 3].

* **Hallazgos principales:**
  1. Tendencia de depreciación (Caída de precio): Se observa una clara inclinación hacia abajo. A medida que los puntos se desplazan hacia la derecha (más millas recorridas), la altura en el eje Y baja (menor precio pedido).
  2. El núcleo del mercado: La mayor cantidad de puntos forma una nube densa en la esquina inferior izquierda. Esto nos dice que el mercado común está compuesto por autos con menos de 200,000 millas (`0.2M`) y precios por debajo de $50,000 dólares.
  3. Puntos atípicos (Casos especiales):
     * **Arriba (Precios muy altos):** Puntos aislados con precios cercanos a $300,000 o $375,000 dólares (posibles vehículos comerciales o errores de captura al publicar).
     * **A la extrema derecha (Kilometraje extremo):** Puntos solitarios cerca de 1,000,000 de millas (`1M`), que corresponden a autos con uso desmedido o números arbitrarios colocados por el vendedor.

* Podríoamos concluir que el desgaste por uso (kilometraje) es un factor determinante en el valor del automóvil: a mayor recorrido, menor es su costo en el mercado.
